```python
#importing all neccesory libraries

import pandas as pd
import numpy as np                                                                   
import plotly.express as px
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('creditcard.csv') #helps import the dataset 
```

```python
df.head(10) #shows us the first 10 rows of the dataset 
```

```python
df['Class'].value_counts(normalize=True)*100 #shows us the imbalance in data
```

```python
df.info() #helps us find the data types of differernt categories
```

```python
df.describe() #helps us in finding the stats of the dataset
```

```python
df.isnull().sum() #no missing values in the dataset
```

```python
fig=px.histogram(df,x='Class',title="fraud vs legit") #helps us visualize the imbalance
fig.show()
```

```python
#the input and target column
X=df.drop("Class",axis=1) 
y=df["Class"]
```

```python
#handles the imabalance of data by creating fake rows for the mdeol
smote=SMOTE()
X_resampled,y_resampled=smote.fit_resample(X,y)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_resampled,y_resampled,test_size=0.2,random_state=42) #divides the data into training and testing sets
```

```python
lr=LogisticRegression(max_iter=1000,solver='saga') #Logisitc regression model
lr.fit(X_train,y_train)
y_pred_lr=lr.predict(X_test)
print("Logistic Regression:")
print("Accuracy score", accuracy_score(y_test,y_pred_lr))
print("Classification Report",classification_report(y_test,y_pred_lr))
```

```python
dt=DecisionTreeClassifier() #Decision tree model
dt.fit(X_train,y_train)
y_pred_dt=dt.predict(X_test)
print("Decision Tree classifier:")
print("accuracy_score",accuracy_score(y_test,y_pred_dt))
print ("classification report",classification_report(y_test,y_pred_dt))
```

```python
rf=RandomForestClassifier() #Random Forest model
rf.fit(X_train,y_train)
y_pred_rf=rf.predict(X_test)
print("accuracy_score",accuracy_score(y_test,y_pred_rf))
print(classification_report,classification_report(y_test,y_pred_rf))
```

```python
cm=confusion_matrix(y_test,y_pred_rf) #confusion matrix tells us whhere the model was right aand wher it went wrong
fig=px.imshow(cm,text_auto=True,title="confusion matrix of Random Forest",labels=dict(x="predicted",y="actual"))
fig.show()
```

```python
feat_imp = pd.Series(rf.feature_importances_, index=X.columns) #tells us all the important categoires model used for prediction
fig = px.bar(feat_imp.sort_values(ascending=False), title='Feature Importance')
fig.show()
```
